```python
square = lambda x: x * x
print(square(5))
```

```python
multiply = lambda a, b: a * b
print(multiply(4,5))
```

```python
check = lambda x: "Even" if x % 2 == 0 else "Odd"
print(check(8))
```

```python
print((lambda a, b: a + b)(10, 20))
```

```python
#lambda with list
numbers = [1, 2, 3, 4, 5]
result = list(map(lambda x: x * 2, numbers))
print(result)
```

```python
numbers = [1, 2, 3, 4]
result = list(map(lambda x: x + 10, numbers))
print(result)
```

```python
numbers = [1, 2, 3, 4, 5, 6]
result = list(filter(lambda x: x % 2 == 0, numbers)) # check condition and return only true values
print(result)
```

```python
names = ["Ali", "Ahmed", "Areeha", "Sara"]
result = sorted(names, key=lambda x: len(x))
print(result)
```

```python
students = [
    {"name": "Ali", "marks": 70},
    {"name": "Sara", "marks": 90},
    {"name": "Ahmed", "marks": 80}
]
result = sorted(students, key=lambda x: x["marks"])
print(result)
```

```python
#without lambda
def square(x):
    return x * x
numbers = [1, 2, 3, 4]
result = list(map(square, numbers))

print(result)
```
